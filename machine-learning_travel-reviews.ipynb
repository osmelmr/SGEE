## 📝 2. Introducción

El presente informe aborda la tarea de **descubrimiento de conocimiento en bases de datos (KDD)** aplicando técnicas de **Aprendizaje Automático No Supervisado** sobre un conjunto de datos real para segmentar preferencias de viajeros.

---

### 2.1. Descripción del Conjunto de Datos

El conjunto de datos seleccionado es **"Travel Reviews"** (Reseñas de Viajes), proveniente del repositorio *UCI Machine Learning* (Dheeru & Casey, 2017). El dataset está basado en **980 reseñas de usuarios** obtenidas de la plataforma TripAdvisor, centradas en la experiencia de viaje en destinos de Asia Oriental.

El dataset se compone de 11 atributos: una columna de identificación de usuario y **10 características predictivas** que representan las **calificaciones promedio (escala de 0 a 4)** que cada usuario otorgó a diferentes categorías de lugares de interés, tales como: *Galerías de arte, Clubes de baile, Restaurantes, Museos, Playas, Resorts, Parques, Teatros*, entre otros.

---

### 2.2. Breve Descripción del Problema a Resolver

El problema abordado es de **Segmentación de Clientes** o **Agrupamiento (Clustering)**, enmarcado en el **Aprendizaje No Supervisado**.

Dado que el dataset no posee una variable objetivo o etiquetas predefinidas, el objetivo es **descubrir patrones y estructuras inherentes** en las preferencias de los viajeros. El análisis busca identificar **grupos naturales** de usuarios que compartan gustos similares (por ejemplo, aquellos que valoran altamente Museos y bajo las Playas) para que una entidad turística pueda optimizar sus estrategias de marketing y recomendaciones.

---

### 2.3. Algoritmos de Agrupamiento a Utilizar

Se seleccionaron tres algoritmos representativos de las tres principales filosofías de *clustering* para realizar una comparación robusta: **K-Means**, **Clustering Jerárquico Aglomerativo** y **DBSCAN**.

#### Resumen de Características de Algoritmos

| Algoritmo | Característica Principal |
| :--- | :--- |
| **K-Means** | **Basado en Centroides** (calcula el promedio del grupo). |
| **Clustering Jerárquico Aglomerativo** | **Basado en Jerarquía** (construye un árbol de relaciones). |
| **DBSCAN** | **Basado en Densidad** (busca concentración de puntos). |

#### Detalle de Algoritmos

##### 1. K-Means
* **Ventajas:** Es el algoritmo estándar, rápido y eficiente. Fácil de implementar e interpretar.
* **Desventajas:** Requiere que el número de grupos ($K$) se defina *a priori*. Es sensible a *outliers* y a la escala de los datos.
* **Aplicaciones Típicas:** Segmentación de mercado, compresión de datos, optimización logística.

##### 2. Clustering Jerárquico Aglomerativo
* **Ventajas:** No requiere definir $K$ al inicio, mostrando todas las posibles particiones (dendrograma). Útil para visualizar subgrupos anidados.
* **Desventajas:** Lento y computacionalmente costoso para datasets muy grandes. Una mala fusión al inicio no se puede revertir.
* **Aplicaciones Típicas:** Biología (taxonomía), análisis de estructuras moleculares.

##### 3. DBSCAN
* **Ventajas:** Capaz de encontrar clústeres de formas irregulares. Excelente para identificar y etiquetar el **ruido** (*outliers*).
* **Desventajas:** La calidad del resultado depende de una correcta definición de sus dos parámetros ($\epsilon$ y $MinPts$). Le cuesta trabajar con clústeres de densidades muy diferentes.
* **Aplicaciones Típicas:** Detección de fraude y anomalías, análisis espacial de datos.
